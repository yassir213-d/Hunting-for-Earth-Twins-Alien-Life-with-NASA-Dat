```python
# ==============================================================================
# 🌌 Project: Searching for Earth Twins & Intelligent Life Using NASA Data
# 🌍 Author: Yassir Dev
# 📡 Data Source: NASA Exoplanet Archive (pscomppars table)
# ==============================================================================

import pandas as pd
import numpy as np

print("📡 Connecting to NASA Exoplanet Archive... Scanning the cosmos...")

# Direct URL to fetch confirmed exoplanets directly in CSV format from NASA's API
url = "https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query=select+pl_name,pl_rade,pl_eqt,pl_insol+from+pscomppars&format=csv"

try:
    # 1. Data Ingestion and Cleaning
    nasa_data = pd.read_csv(url)
    
    # Drop rows missing critical metrics (Planet Name, Radius, or Equilibrium Temp)
    nasa_data = nasa_data.dropna(subset=['pl_name', 'pl_rade', 'pl_eqt'])
    
    print(f"✅ Connection Successful! Downloaded {len(nasa_data):,} planets from NASA.\n")
    
    # 2. Scientific Filtering (The Habitable Goldilocks Zone)
    # pl_rade: Planet radius between 0.5 and 1.5 of Earth's radius (Rocky planets)
    # pl_eqt: Equilibrium temperature between 180K and 310K (Allows for liquid water and atmosphere)
    earth_like = nasa_data[
        (nasa_data['pl_rade'] >= 0.5) & (nasa_data['pl_rade'] <= 1.5) & 
        (nasa_data['pl_eqt'] >= 180) & (nasa_data['pl_eqt'] <= 310)
    ]
    
    print(f"🌍 Number of Earth-like planets found: {len(earth_like)}")
    
    if len(earth_like) > 0:
        print("\n📜 Top Candidate Planets for Potential Intelligent Life:")
        print("=" * 65)
        # Displaying the filtered results cleanly
        print(earth_like[['pl_name', 'pl_rade', 'pl_eqt']].to_string(index=False))
        print("=" * 65)
        
        # 3. Statistical Probability of Intelligent Life (Simplified Drake Equation)
        # Assumptions: 10% of habitable planets develop life, and 1% of those evolve tech civilizations
        total_candidates = len(earth_like)
        prob_life = total_candidates * 0.10 * 0.01
        print(f"\n🔮 Statistical Probability of Intelligent Civilizations in this sample: ~{prob_life:.4f}")
        
    else:
        print("\n🤔 Strict filters! No exact Earth twin in this batch. Try to expand parameters.")

except Exception as e:
    print(f"❌ Error occurred: {e}")
    print("If the server is busy, try running the cell again in a few moments.")
```
